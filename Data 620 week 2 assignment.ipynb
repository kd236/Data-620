# Hello, Graph World

This notebook creates a simple network graph using NetworkX in Python.
The nodes represent hospital departments, and the edges represent communication or collaboration between departments.


```python
import networkx as nx
import matplotlib.pyplot as plt

# Create an undirected graph
G = nx.Graph()

# Add nodes
G.add_nodes_from([
    "Quality",
    "Nursing",
    "Physicians",
    "Pharmacy",
    "IT",
    "Administration"
])

# Add edges
G.add_edges_from([
    ("Quality", "Nursing"),
    ("Quality", "Physicians"),
    ("Quality", "Administration"),
    ("Nursing", "Physicians"),
    ("Physicians", "Pharmacy"),
    ("Pharmacy", "IT"),
    ("IT", "Administration")
])

# Draw the graph
plt.figure(figsize=(8, 6))

nx.draw(
    G,
    with_labels=True,
    node_size=2500,
    font_size=10
)

plt.title("Hospital Department Communication Network")
plt.show()
```
